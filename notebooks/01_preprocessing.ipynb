```python
import math
import os

from msi_vae.data import *
```

```python
file_name = "5 June tumour test 2_1-327482_SN0p0_profile.h5"

dataset_directory = os.path.join(
    os.environ["HOME"],
    ".datasets",
    "mass_spec_data",
    "Cancer biopsy",
    "5 June",
)

file_path = os.path.join(dataset_directory, file_name)

if not os.path.exists(file_path):
    raise Exception(f"Please add the h5 file to the path: {file_path}")
```

```python
f = read_file(file_path)
```

```python
spectrum_ids = list(int(key) for key in f.keys())
sorted_spectrum_ids = sorted(spectrum_ids)
```

```python
print(f"Number of spectra: {len(sorted_spectrum_ids)}")
print(f"Lowest spectra index: {min(sorted_spectrum_ids)}")
print(f"Highest spectra index: {max(sorted_spectrum_ids)}")
```

```python
mz_min, mz_max = get_min_and_max_mass_to_charge(f)
```

```python
print(f"Minimum mass-to-charge across spectra: {mz_min}")
print(f"Maximum mass-to-charge across spectra: {mz_max}")
```

```python
bin_width = 0.02
bins = np.arange(math.floor(mz_min), math.ceil(mz_max) + bin_width, bin_width)
print("Number of bins:", len(bins) - 1)
```

```python
processed_file = os.path.join(dataset_directory, "binned_with_tic.h5")

if os.path.exists(processed_file):
    print(f"The file already exists")
else:
    print(f"Saving processed file to: {processed_file}")
    bin_spectra(f, sorted_spectrum_ids, bins, processed_file)
```
